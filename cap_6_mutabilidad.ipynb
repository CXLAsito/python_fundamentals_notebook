### Los objetos integer no son mutables:

```python
a = 3                  # variables de integers no son mutables, su valor no se verá alterado al relacionarla a otra variable que se cambia.
b = a
a = a + 2              # b es 3
print("b es ", b)
```

# OBJETOS MUTABLES

### Listas

```python
L = [2, 3, 4]	    # A mutable object
M = L			    # Make a reference to the same object
L[0] = 24 		    # An in-place change

print("L = ", L)    # [24, 3, 4]
print("M = ", M)    # [24, 3, 4]
```

```python
L = [2, 3, 4]	    # A mutable object
M = L			    # Make a reference to the same object
M[0] = 24 		    # An in-place change

print("L = ", L)    # [24, 3, 4]
print("M = ", M)    # [24, 3, 4]
```

### Shared References and Equality

- Operador ``==`` igualdad de valores.
- Operador ``is`` identidad de objetos.

```python
L = [1, 2, 3]
M = L			    # M and L reference the same object

L == M 			    # True: Same values => operador igualdad de valores
L is M 			    # True: operador identidad de objetos: compara las referencias (los punteros)
```

```python
L = [1, 2, 3]
M = [1, 2, 3]			        # M and L reference the same object

L == M 			    # True: Same values => operador igualdad de valores
L is M 			    # False: operador identidad de objetos: compara las referencias (los punteros)
```

```python
L = [1, 2, 3]
M = [1, 2, 3]	    # M and L reference different objects

print(
    "L == M",
    L == M 			# True: Same values
)

print(
    "L is M", 	    # False: operador identidad de objetos: compara las referencias (los punteros)
    L is M
)
```

### CACHE
Because small integers and strings are cached and reused, though, it tells us they reference the same single object.

```python
X = 42
Y = 42		    # Should be two different objects

X  == Y         # True
X is Y 		    # True: Same object anyhow (does not require X = Y): caching at work!!
```
